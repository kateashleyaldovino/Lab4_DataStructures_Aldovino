```python

LAST_NAME = "Aldovino"
SEED_NUM = 8
FAVORITE_ARTIST = "Taylor Swift"

print("Student:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
```

```python

from collections import Counter, defaultdict
from statistics import mean

# 1. Generate engineering test data
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

artist_letters = [ord(c.lower()) - ord('a') + 1
                  for c in FAVORITE_ARTIST.lower() if c.isalpha()]

if not artist_letters:
    artist_letters = [1]

test_records = []

for i, source in enumerate(sources):
    for j, category in enumerate(categories):
        value = SEED_NUM + (artist_letters[(i + j) % len(artist_letters)]) + (i * 5) + j
        test_records.append({
            "source": source,
            "category": category,
            "value": value
        })

# Add a repeated test-source/category record
test_records.append(test_records[0].copy())

# Add multiple measurements under selected categories
test_records.append({
    "source": sources[0],
    "category": "TEMPERATURE",
    "value": SEED_NUM + 10
})
test_records.append({
    "source": sources[1],
    "category": "VOLTAGE",
    "value": SEED_NUM + 20
})

# 2. Organize data by source and category
organized = defaultdict(lambda: defaultdict(list))

for record in test_records:
    organized[record["source"]][record["category"]].append(record["value"])

print("=== GENERATED TEST DATA ===")
for record in test_records:
    print(record)

print("\n=== ORGANIZED TEST DATA ===")
for source, measurements in organized.items():
    print(source, dict(measurements))

# 3. Identify repeated and distinct records
record_counts = Counter(
    (r["source"], r["category"], r["value"]) for r in test_records
)

repeated_records = {k: v for k, v in record_counts.items() if v > 1}
distinct_records = set(
    (r["source"], r["category"], r["value"]) for r in test_records
)

print("\n=== REPEATED RECORDS ===")
print(repeated_records)

print("\n=== DISTINCT RECORD COUNT ===")
print(len(distinct_records))

# 4. Numerical and categorical summaries
print("\n=== CATEGORY SUMMARIES ===")
for category in categories:
    values = [r["value"] for r in test_records
              if r["category"] == category]
    print(category, {
        "count": len(values),
        "mean": round(mean(values), 2),
        "minimum": min(values),
        "maximum": max(values)
    })

print("\n=== SOURCE SUMMARIES ===")
for source in sources:
    values = [r["value"] for r in test_records
              if r["source"] == source]
    print(source, {
        "measurement_count": len(values),
        "total": sum(values),
        "average": round(mean(values), 2)
    })

# 5. Arrange processed results
ordered_records = sorted(
    test_records,
    key=lambda r: (r["category"], r["source"], r["value"])
)

print("\n=== ORDERED RESULTS ===")
for record in ordered_records:
    print(record)

# 6. Final summary
all_values = [r["value"] for r in test_records]

print("\n=== FINAL SUMMARY ===")
print("Total records:", len(test_records))
print("Distinct records:", len(distinct_records))
print("Repeated record types:", len(repeated_records))
print("Categories:", len(set(r["category"] for r in test_records)))
print("Overall average:", round(mean(all_values), 2))
print("Overall minimum:", min(all_values))
print("Overall maximum:", max(all_values))
```

```python

from collections import Counter, defaultdict
from statistics import mean

# 1. Generate student names and subjects
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

artist_letters = [ord(c.lower()) - ord('a') + 1
                  for c in FAVORITE_ARTIST.lower() if c.isalpha()]

if not artist_letters:
    artist_letters = [1]

performance_records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        score = 50 + (
            SEED_NUM + artist_letters[(i + j) % len(artist_letters)]
            + i * 3 + j * 2
        ) % 51

        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

# 2. Add one repeated student-subject record
performance_records.append(performance_records[0].copy())

# Add one incomplete record with a missing score
performance_records.append({
    "student": students[0],
    "subject": "MATHEMATICS",
    "score": None
})

# 3. Determine performance categories
def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 75:
        return "PASSED"
    else:
        return "FAILED"

for record in performance_records:
    record["category"] = performance_category(record["score"])

print("=== GENERATED PERFORMANCE RECORDS ===")
for record in performance_records:
    print(record)

# 4. Organize records by student and subject
organized = defaultdict(lambda: defaultdict(list))

for record in performance_records:
    organized[record["student"]][record["subject"]].append(
        record["score"]
    )

print("\n=== ORGANIZED STUDENT/SUBJECT DATA ===")
for student, subject_data in organized.items():
    print(student, dict(subject_data))

# 5. Identify repeated and incomplete records
record_counts = Counter(
    (r["student"], r["subject"], r["score"])
    for r in performance_records
)

repeated_records = {k: v for k, v in record_counts.items() if v > 1}
incomplete_records = [
    r for r in performance_records if r["score"] is None
]

print("\n=== REPEATED RECORDS ===")
print(repeated_records)

print("\n=== INCOMPLETE RECORDS ===")
print(incomplete_records)

# 6. Student summaries
print("\n=== STUDENT SUMMARIES ===")
for student in students:
    scores = [
        r["score"] for r in performance_records
        if r["student"] == student and r["score"] is not None
    ]

    if scores:
        print(student, {
            "valid_scores": len(scores),
            "average": round(mean(scores), 2),
            "highest": max(scores),
            "lowest": min(scores)
        })

# 7. Subject summaries
print("\n=== SUBJECT SUMMARIES ===")
for subject in subjects:
    scores = [
        r["score"] for r in performance_records
        if r["subject"] == subject and r["score"] is not None
    ]

    if scores:
        print(subject, {
            "valid_scores": len(scores),
            "average": round(mean(scores), 2),
            "highest": max(scores),
            "lowest": min(scores)
        })

# 8. Performance category counts
category_counts = Counter(
    r["category"] for r in performance_records
)

print("\n=== PERFORMANCE CATEGORIES ===")
print(dict(category_counts))

# 9. Ordered results
ordered_records = sorted(
    performance_records,
    key=lambda r: (
        r["student"],
        r["subject"],
        r["score"] is None,
        r["score"] if r["score"] is not None else 0
    )
)

print("\n=== ORDERED RESULTS ===")
for record in ordered_records:
    print(record)

# 10. Final report
valid_scores = [
    r["score"] for r in performance_records
    if r["score"] is not None
]

print("\n=== FINAL STUDENT PERFORMANCE REPORT ===")
print("Total records:", len(performance_records))
print("Repeated record types:", len(repeated_records))
print("Incomplete records:", len(incomplete_records))
print("Valid scores:", len(valid_scores))
print("Overall average:", round(mean(valid_scores), 2))
print("Highest score:", max(valid_scores))
print("Lowest score:", min(valid_scores))
```

```python

from collections import Counter, defaultdict

# 1. Generate customers and products
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

prices = {
    "NOTEBOOK": 35,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 350,
    "CALCULATOR": 500
}

products = list(prices.keys())

artist_letters = [ord(c.lower()) - ord('a') + 1
                  for c in FAVORITE_ARTIST.lower() if c.isalpha()]

if not artist_letters:
    artist_letters = [1]

# 2. Generate multiple transactions
transactions = []

for i, customer in enumerate(customers):
    for j, product in enumerate(products[:3]):
        quantity = 1 + (
            SEED_NUM + artist_letters[(i + j) % len(artist_letters)]
        ) % 5

        transactions.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "price": prices[product]
        })

# Add transactions for the remaining products
for i, product in enumerate(products[3:]):
    transactions.append({
        "customer": customers[i],
        "product": product,
        "quantity": 1 + (SEED_NUM + i) % 5,
        "price": prices[product]
    })

# Add a repeated customer-product purchase
transactions.append(transactions[0].copy())

# 3. Calculate transaction values
for transaction in transactions:
    transaction["total"] = (
        transaction["quantity"] * transaction["price"]
    )

print("=== GENERATED TRANSACTION RECORDS ===")
for transaction in transactions:
    print(transaction)

# 4. Organize transactions
customer_data = defaultdict(list)
product_data = defaultdict(list)

for transaction in transactions:
    customer_data[transaction["customer"]].append(transaction)
    product_data[transaction["product"]].append(transaction)

print("\n=== ORGANIZED TRANSACTION DATA ===")
for customer, records in customer_data.items():
    print(customer, records)

# 5. Identify repeated purchases and distinct data
purchase_counts = Counter(
    (t["customer"], t["product"]) for t in transactions
)

repeated_purchases = {
    purchase: count
    for purchase, count in purchase_counts.items()
    if count > 1
}

print("\n=== REPEATED CUSTOMER-PRODUCT PURCHASES ===")
print(repeated_purchases)

print("\n=== DISTINCT CUSTOMERS ===")
print(len(set(t["customer"] for t in transactions)))

print("\n=== DISTINCT PRODUCTS ===")
print(len(set(t["product"] for t in transactions)))

# 6. Transaction values and customer summaries
print("\n=== CUSTOMER SUMMARIES ===")
for customer in customers:
    records = customer_data[customer]
    print(customer, {
        "transaction_count": len(records),
        "total_quantity": sum(t["quantity"] for t in records),
        "total_spending": sum(t["total"] for t in records)
    })

# 7. Product summaries
print("\n=== PRODUCT SUMMARIES ===")
for product in products:
    records = product_data[product]

    if records:
        print(product, {
            "price": prices[product],
            "units_sold": sum(t["quantity"] for t in records),
            "transaction_count": len(records),
            "total_sales": sum(t["total"] for t in records)
        })
    else:
        print(product, {
            "price": prices[product],
            "units_sold": 0,
            "transaction_count": 0,
            "total_sales": 0
        })

# 8. Ordered results
ordered_transactions = sorted(
    transactions,
    key=lambda t: (t["customer"], t["product"])
)

print("\n=== ORDERED TRANSACTIONS ===")
for transaction in ordered_transactions:
    print(transaction)

# 9. Overall transaction summary
total_quantity = sum(t["quantity"] for t in transactions)
total_sales = sum(t["total"] for t in transactions)

print("\n=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print("Total transaction records:", len(transactions))
print("Repeated purchase types:", len(repeated_purchases))
print("Distinct customers:", len(set(t["customer"] for t in transactions)))
print("Distinct products:", len(set(t["product"] for t in transactions)))
print("Total units sold:", total_quantity)
print("Overall sales: PHP", f"{total_sales:,.2f}")
print("Average transaction value: PHP",
      f"{total_sales / len(transactions):,.2f}")
```
